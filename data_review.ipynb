```python
import pandas as pd
from pathlib import Path
from src.compare_datasets import compare_datasets
```

```python
DATA_DIR = Path("./data")
TDWC0101_DIR = DATA_DIR / "TDWC0101"

TDWC0101_FIS_RAW_FILE = TDWC0101_DIR / "raw" / "TDWC0101_FIS.ods"
TDWC0101_TMS_RAW_FILE = TDWC0101_DIR / "raw" / "TDWC0101_TMS.csv"

df_fis = pd.read_excel(TDWC0101_FIS_RAW_FILE, engine="odf", sheet_name="data", header=1, dtype=str)
df_tms = pd.read_csv(TDWC0101_TMS_RAW_FILE, dtype=str)

print(f"[INFO] FIS columns: {list(df_fis.columns)}")
print(f"[INFO] TMS columns: {list(df_tms.columns)}")
```

```python
# Strip trailing spaces from TMS column names
df_tms.columns = df_tms.columns.str.strip()

# Rename FIS columns to match TMS
df_fis = df_fis.rename(columns={
    'Locode':            'Start Locode',
    'Start':             'Start Location',
    'Locode.1':          'Pos/Del Locode',
    'Pos/Del':           'Pos/Del Location',
    'Locode.2':          'End Locode',
    'End':               'End Location',
    'Date':              'Pos/Del Date',
    'Time.1':            'Pos/Del Time',
    'Locode.3':          'Multistop Locode',
    'Multistop':         'Multistop Location',
    'OW Surcharge':      'Overweight Surcharge',
    'MautToll Surcharge':'Maut/Toll Surcharge',
}).drop(columns=['Unnamed: 19'], errors='ignore')

print(f"[INFO] FIS columns: {list(df_fis.columns)}")
print(f"[INFO] TMS columns: {list(df_tms.columns)}")
```

```python
summary = compare_datasets(
    df_fis, df_tms,
    join_cols=["Number", "Cont. Number"],
    output_dir="data/TDWC0101/dev",
)
summary
```
